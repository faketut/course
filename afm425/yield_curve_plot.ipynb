```python
import matplotlib.pyplot as plt
import numpy as np

# Tenors and ordinal positions for equal visual spacing
tenors = ['1M', '2M', '3M', '4M', '6M', '1Y', '2Y', '3Y', '5Y', '7Y', '10Y', '20Y', '30Y']
x = np.arange(len(tenors))

# Yield curve data
curves = {
    '09/15/2008 (GFC)': [0.36, np.nan, 1.02, np.nan, 1.55, 1.66, 1.78, 2.01, 2.59, 2.99, 3.47, 4.14, 4.12],
    '01/31/2020 (Pre-COVID)': [1.56, 1.57, 1.55, np.nan, 1.54, 1.45, 1.33, 1.30, 1.32, 1.42, 1.51, 1.83, 1.99],
    '04/20/2020 (COVID Panic)': [0.10, 0.10, 0.12, np.nan, 0.15, 0.15, 0.20, 0.26, 0.35, 0.51, 0.63, 1.04, 1.23],
    '02/28/2022 (Pre-Hikes)': [0.06, 0.20, 0.35, np.nan, 0.69, 1.01, 1.44, 1.62, 1.71, 1.81, 1.83, 2.25, 2.17],
    '07/03/2023 (Inverted)': [5.27, 5.40, 5.44, 5.52, 5.53, 5.43, 4.94, 4.56, 4.19, 4.03, 3.86, 4.08, 3.87],
    '09/23/2026 (Current)': [3.99, 4.07, 4.10, 4.19, 4.30, 4.31, 4.49, 4.85, 4.97, 4.99, 5.05, 5.11, 5.40]
}

colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']

plt.figure(figsize=(12, 6))

for (label, y_vals), color in zip(curves.items(), colors):
    # Mask out NaN values to connect lines cleanly
    y_arr = np.array(y_vals)
    mask = ~np.isnan(y_arr)
    plt.plot(x[mask], y_arr[mask], marker='o', linewidth=2, label=label, color=color)

plt.xticks(x, tenors)
plt.title('U.S. Treasury Yield Curves Across Representative Periods (AFM 425 Case #4)', fontsize=13, fontweight='bold')
plt.xlabel('Maturity Tenor', fontsize=11)
plt.ylabel('Yield (%)', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(title='Selected Dates', loc='upper left')
plt.tight_layout()
plt.show()
```
